IMPORT MODULE FROM GEMINI PACKAGE

```python
from google import genai
```

```python
from google.colab import userdata
```

USE CLIENT CLASS TO CREATE AN INSTANCE (FOR INTERACTING WITH GEMIN API)

```python
client = genai.Client(api_key = userdata.get("GEMINI_API_KEY"))
client
```

USE CLIENT OBJECT FOR COMMUNICATING TO GEMINI LLM

```python
def study_assistant(user_prompt):
  response = client.models.generate_content(
      model = 'gemini-3.6-flash',
      contents = user_prompt
  )
  return response
```

```python
output = study_assistant("Explain LLM architecture")
print(output)
```
